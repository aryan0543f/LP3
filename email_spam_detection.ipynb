# Email Spam Detection
**Binary Classification** — Normal State (Not Spam) vs Abnormal State (Spam)  
**Models:** K-Nearest Neighbors (KNN) | Support Vector Machine (SVM)

---

## Objective
Classify emails as spam or legitimate (ham) using machine learning classification algorithms.

## Import Required Libraries

````python
# Data manipulation and analysis
import pandas as pd

# Visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns
````

## STEP 1 – Load Dataset
Load the email dataset which contains word frequency counts as features.

````python
# Load CSV file containing email data
# Each column represents the frequency of a specific word in an email
# Last column 'Prediction' contains labels: 0 = Not Spam, 1 = Spam
df = pd.read_csv('emails.csv')

# Display first 5 rows to understand data structure
df.head()
````

## STEP 2 – Check for Missing Values
Identify any missing (null) values in the dataset that need to be handled.

````python
# Count null values in each column
# A clean dataset should have 0 null values
df.isnull().sum()
````

## STEP 3 – Drop Null Rows & Prepare Features / Target
- Skip `Email No.` (column 0) — it is just an identifier
- `X` = word frequency columns (index 1 to second-last)
- `y` = last column `Prediction` (0 = Not Spam, 1 = Spam)

````python
# Remove any rows with missing values
df.dropna(how='any', inplace=True)

# Separate features (X) and target (y)
# iloc[:, 1:-1] selects all rows, columns from index 1 to second-last (word frequencies)
x = df.iloc[:, 1:-1].values

# iloc[:, -1] selects the last column (Prediction: 0 or 1)
y = df.iloc[:, -1].values
````

## STEP 4 – Train / Test Split
Divide the dataset into training (75%) and testing (25%) sets.

````python
from sklearn.model_selection import train_test_split

# Split data: 75% training, 25% testing
# random_state=10 ensures reproducible results
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=10
)
````

## STEP 5 – Report Helper Function
Creates a reusable function to evaluate and visualize model performance.

**Metrics:**
- **Accuracy**: Overall correctness (TP + TN) / Total
- **Precision**: Of predicted spam, how many are actually spam? TP / (TP + FP)
- **Recall**: Of actual spam, how many did we catch? TP / (TP + FN)

**Visualizations:**
- **Confusion Matrix**: Shows TP, TN, FP, FN
- **Precision-Recall Curve**: Trade-off between precision and recall
- **ROC Curve**: True Positive Rate vs False Positive Rate

````python
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    accuracy_score, precision_score, recall_score,
    PrecisionRecallDisplay, RocCurveDisplay
)

def report(classifier):
    """
    Generate comprehensive evaluation report for a classifier.
    
    Args:
        classifier: Trained sklearn classifier with predict() method
    
    Displays:
        - Confusion Matrix
        - Accuracy, Precision, Recall scores
        - Precision-Recall Curve
        - ROC Curve
    """
    # Make predictions on test set
    y_pred = classifier.predict(x_test)

    # 1. Confusion Matrix
    # Shows: [[TN, FP], [FN, TP]]
    cm = confusion_matrix(y_test, y_pred)
    ConfusionMatrixDisplay(cm, display_labels=classifier.classes_).plot()
    plt.show()

    # 2. Performance Metrics
    print(f"Accuracy:        {accuracy_score(y_test, y_pred)}")
    print(f"Precision Score: {precision_score(y_test, y_pred)}")
    print(f"Recall Score:    {recall_score(y_test, y_pred)}")

    # 3. Precision-Recall Curve
    # Shows trade-off between precision and recall at different thresholds
    PrecisionRecallDisplay.from_estimator(classifier, x_test, y_test)
    plt.show()

    # 4. ROC Curve (Receiver Operating Characteristic)
    # Shows True Positive Rate vs False Positive Rate
    # AUC (Area Under Curve) closer to 1.0 = better model
    RocCurveDisplay.from_estimator(classifier, x_test, y_test)
    plt.show()
````

## STEP 6 – K-Nearest Neighbors Classifier

**KNN Algorithm:**
- Instance-based learning (lazy learning)
- Classifies based on majority vote of K nearest neighbors
- Uses distance metrics (Euclidean by default)
- `n_neighbors=10`: Considers 10 closest training examples

````python
from sklearn.neighbors import KNeighborsClassifier

# Initialize KNN classifier with K=10
# Higher K = smoother decision boundaries, less overfitting
# Lower K = more complex boundaries, risk of overfitting
kNN = KNeighborsClassifier(n_neighbors=10)

# Train the model on training data
kNN.fit(x_train, y_train)
````

### Evaluate KNN Performance

````python
# Generate comprehensive evaluation report
report(kNN)
````

## STEP 7 – Support Vector Machine (SVM) Classifier

**SVM Algorithm:**
- Finds optimal hyperplane that maximizes margin between classes
- Effective in high-dimensional spaces (good for text data)
- Uses kernel functions to handle non-linear relationships
- Default kernel: RBF (Radial Basis Function)

````python
from sklearn.svm import SVC

# Initialize SVM classifier
# Default kernel='rbf' handles non-linear decision boundaries
# gamma='scale' automatically adjusts kernel coefficient
svm = SVC()

# Train the model on training data
svm.fit(x_train, y_train)
````

### Evaluate SVM Performance

````python
# Generate comprehensive evaluation report
report(svm)
````

## 📖 Technical Terms Explained

### **Classification Concepts**

#### **1. Binary Classification**
- Dividing data into two categories (classes)
- In our case: Spam (1) vs Not Spam (0)
- Different from multi-class (3+ categories) or regression (continuous values)

#### **2. Supervised Learning**
- Training with labeled data (we know the correct answers)
- Model learns patterns from features → labels
- Opposite: Unsupervised learning (no labels, finds patterns)

#### **3. Features (Independent Variables)**
- Input data used to make predictions
- In our case: Word frequency counts (3000 words)
- Represented as **X** in code

#### **4. Target/Labels (Dependent Variable)**
- What we're trying to predict
- In our case: 0 (Not Spam) or 1 (Spam)
- Represented as **y** in code

#### **5. Training vs Testing**
- **Training set (75%)**: Used to teach the model
- **Testing set (25%)**: Used to evaluate model on unseen data
- *Why split?* Prevents overfitting, gives realistic performance estimate

### **Evaluation Metrics Explained**

#### **Confusion Matrix Components:**
```
                  Predicted
                 Not Spam  Spam
Actual Not Spam    TN      FP
       Spam        FN      TP
```

- **TP (True Positive)**: Correctly identified spam
- **TN (True Negative)**: Correctly identified legitimate email
- **FP (False Positive)**: Legitimate email marked as spam (Type I Error)
- **FN (False Negative)**: Spam email that got through (Type II Error)

#### **1. Accuracy**
- **Formula**: `(TP + TN) / (TP + TN + FP + FN)`
- **Interpretation**: Overall correctness of predictions
- **Range**: 0 to 1 (0% to 100%)
- **Example**: 0.90 accuracy = 90% of predictions are correct
- **Limitation**: Can be misleading with imbalanced datasets

#### **2. Precision**
- **Formula**: `TP / (TP + FP)`
- **Question answered**: "Of emails we marked as spam, how many actually were spam?"
- **High precision**: Few false positives (few legitimate emails blocked)
- **Trade-off**: Might miss some spam (lower recall)
- **Use case**: Important when false positives are costly (e.g., blocking important emails)

#### **3. Recall (Sensitivity/True Positive Rate)**
- **Formula**: `TP / (TP + FN)`
- **Question answered**: "Of all actual spam emails, how many did we catch?"
- **High recall**: Few false negatives (few spam emails get through)
- **Trade-off**: Might flag more legitimate emails as spam (lower precision)
- **Use case**: Important when missing positives is costly (e.g., medical diagnosis)

#### **4. F1-Score** (not directly shown but important)
- **Formula**: `2 × (Precision × Recall) / (Precision + Recall)`
- **Purpose**: Harmonic mean balancing precision and recall
- **Use case**: When you need balance between precision and recall

#### **5. ROC Curve (Receiver Operating Characteristic)**
- **X-axis**: False Positive Rate (FPR) = FP / (FP + TN)
- **Y-axis**: True Positive Rate (TPR) = Recall = TP / (TP + FN)
- **AUC (Area Under Curve)**:
  - 1.0 = Perfect classifier
  - 0.9-1.0 = Excellent
  - 0.8-0.9 = Good
  - 0.7-0.8 = Acceptable
  - 0.5 = Random guess (diagonal line)
- **Interpretation**: Model's ability to distinguish between classes

#### **6. Precision-Recall Curve**
- Shows trade-off between precision and recall at different thresholds
- **High threshold**: High precision, low recall (strict about calling spam)
- **Low threshold**: Low precision, high recall (aggressive spam detection)
- **Use case**: Choose threshold based on business requirements

### **Algorithms Used**

#### **1. K-Nearest Neighbors (KNN)**

**Type**: Instance-based, non-parametric, lazy learning

**How it works:**
1. Store all training data (doesn't build a model upfront)
2. For new email, find K nearest training examples
3. Use majority vote of those K neighbors
4. Distance metric: Euclidean distance by default

**Key Parameter:**
- **n_neighbors (K)**: Number of neighbors to consider
  - K=1: Uses only closest neighbor (prone to noise)
  - K=10 (our choice): Balanced, smoother decision boundaries
  - K=100: Very smooth, might miss patterns
  - **Rule of thumb**: K = √n (where n = number of samples)

**Distance Calculation (Euclidean):**
```
distance = √[(x₁-x₂)² + (y₁-y₂)² + ... + (zₙ-zₙ)²]
```
For 3000 features, it's 3000-dimensional distance!

**Pros:**
- Simple to understand and implement
- No training phase (just stores data)
- Works well with multi-class problems
- Non-linear decision boundaries

**Cons:**
- Slow prediction (must calculate distance to all training points)
- Memory intensive (stores all training data)
- Sensitive to feature scaling
- Curse of dimensionality (struggles with many features)
- Affected by irrelevant features

**When to use:**
- Small to medium datasets
- Need interpretable results
- Non-linear relationships

---

#### **2. Support Vector Machine (SVM)**

**Type**: Supervised, discriminative, kernel-based

**Core Concept:**
- Finds optimal hyperplane (decision boundary) that maximizes margin
- **Margin**: Distance between hyperplane and nearest data points (support vectors)
- **Support Vectors**: Critical data points closest to decision boundary

**How it works:**
1. Find hyperplane that separates classes
2. Maximize margin (distance to nearest points)
3. Use kernel trick for non-linear problems
4. Only support vectors matter (sparse solution)

**Key Parameters:**

1. **kernel**: Transformation function
   - **'linear'**: Straight line separation (fast, interpretable)
   - **'rbf'** (Radial Basis Function - default): Handles non-linear data
     - Formula: `K(x,y) = exp(-gamma × ||x-y||²)`
     - Creates circular/elliptical boundaries
   - **'poly'**: Polynomial kernel
   - **'sigmoid'**: Sigmoid kernel

2. **C** (Regularization parameter):
   - **Low C**: Wider margin, more errors allowed (underfitting)
   - **High C**: Narrower margin, fewer errors (overfitting risk)
   - Default: 1.0

3. **gamma** (Kernel coefficient for RBF):
   - **Low gamma**: Far points have influence (smooth boundary)
   - **High gamma**: Only nearby points matter (complex boundary)
   - **'scale'** (default): 1 / (n_features × X.var())
   - **'auto'**: 1 / n_features

**Kernel Trick:**
- Transforms data to higher dimension without computing it explicitly
- Makes non-linearly separable data → linearly separable
- Example: 2D circle → 3D cone (separable by plane)

**Pros:**
- Effective in high-dimensional spaces (great for text classification)
- Memory efficient (only uses support vectors)
- Versatile (different kernels for different data)
- Works well with clear margin of separation
- Robust to overfitting (especially in high dimensions)

**Cons:**
- Slow training on large datasets (O(n²) to O(n³))
- Requires feature scaling
- Sensitive to hyperparameter choices
- Less interpretable (black box)
- Doesn't provide probability estimates directly

**When to use:**
- High-dimensional data (like text)
- Clear margin between classes
- Medium-sized datasets
- Need for robust performance
